```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}
cart = {}
while True:
    print("\n ONLINE SHOPPING CART")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        print("\n available products")
        print("ID\tProduct\t\tPrice\t\tStock")
        for product_id, details in products.items():
            print(product_id, "\t", details["name"], "\t\tRs.", details["price"], "\t", details["stock"])
    elif choice == "2":
        product_id = input("Enter Product ID: ")
        if product_id in products:
            quantity = int(input("Enter Quantity: "))
            if quantity > 0:
                if product_id in cart:
                    total_quantity = cart[product_id]["quantity"] + quantity
                else:
                    total_quantity = quantity
                if total_quantity <= products[product_id]["stock"]:
                    if product_id in cart:
                        cart[product_id]["quantity"] += quantity
                    else:
                        cart[product_id] = {
                            "quantity": quantity,
                            "price": products[product_id]["price"],
                            "name": products[product_id]["name"]
                        }
                    print("Product added to cart successfully")
                else:
                    print("No stock available")
            else:
                print("Quantity must be greater than zero")
        else:
            print("Invalid Product ID!")
    elif choice == "3":
        if len(cart) == 0:
            print("Your cart is empty")
        else:
            print("\n YOUR SHOPPING CART")
            cart_total = 0
            for product_id, details in cart.items():
                subtotal = details["price"] * details["quantity"]
                cart_total += subtotal
                print("Product:", details["name"])
                print("Quantity:", details["quantity"])
                print("Price: Rs.", details["price"])
                print("Subtotal: Rs.", subtotal)
                print("____________________")
            print("Cart Total: Rs.", cart_total)
    elif choice == "4":
        if len(cart) == 0:
            print("Your cart is empty Add products first.")
        else:
            total = 0
            for product_id, details in cart.items():
                subtotal = details["price"] * details["quantity"]
                total += subtotal
            print("\n FINAL BILL")
            print("Total Amount: Rs.", total)
            if total >= 5000:
                discount = total * 0.10
                final_bill = total - discount
                print("Discount (10%): Rs.", discount)
            else:
                discount = 0
                final_bill = total
                print("Discount: Rs. 0")
            print("Final Bill: Rs.", final_bill)
            for product_id, details in cart.items():
                products[product_id]["stock"] -= details["quantity"]
            cart.clear()
            print("Checkout Successful!")
            print("Cart has been cleared.")
    elif choice == "5":
        print("Thanks for shopping")
        break
    else:
        print("not correct: Please select 1 to 5.")
```
